# Урок 2. НОД, НОК и Линейные комбинации

## НОД и НОК: Базовые свойства

Определение: $\text{НОД}(a, b)$ — наибольшее натуральное число, на которое нацело делятся и $a$, и $b$. $\text{НОК}(a, b)$ — наименьшее число, которое делится и на $a$, и на $b$.

Главное тождество: $\text{НОД}(a, b) \cdot \text{НОК}(a, b) = a \cdot b$.

Например: Для чисел $12$ и $18$: $\text{НОД}(12, 18) = 6$, $\text{НОК}(12, 18) = 36$. Проверяем тождество: $6 \cdot 36 = 216$ и $12 \cdot 18 = 216$.

## Алгоритм Евклида

Суть: Любой общий делитель чисел $a$ и $b$ (где $a > b$) также является делителем их разности $(a - b)$. Это позволяет «срезать» большие числа, не раскладывая их на множители.

Формула: $\text{НОД}(a, b) = \text{НОД}(a - b, b) = \text{НОД}(a \pmod b, b)$.

Например: $\text{НОД}(105, 25) = \text{НОД}(105 - 4 \cdot 25, 25) = \text{НОД}(5, 25) = 5$

## Линейные комбинации выражений

Суть: Если число $d$ делит выражения $A$ и $B$, то оно делит и выражение $\alpha A + \beta B$ при любых целых $\alpha$ и $\beta$. Коэффициенты подбираются крест-накрест, чтобы переменная полностью уничтожилась.

Например: На что может сократиться дробь $\frac{3n+1}{2n+3}$? Умножим числитель на $2$, а знаменатель на $3$: $3(2n+3) - 2(3n+1) = 6n + 9 - 6n - 2 = 7$. Значит, дробь может сократиться только на $7$ (или быть несократимой).

# Урок 6. Линейные и простые диофантовы уравнения

## Линейные диофантовы уравнения ($ax + by = c$)

Суть: Уравнения с двумя неизвестными, решения которых ищутся строго в целых числах ($\mathbb{Z}$). Если $\text{НОД}(a, b)$ не делит $c$, то решений нет. Если делит — решений бесконечно много, и они задаются через целочисленный параметр $k$.

Метод остатков (быстрый перебор): Выражаем переменную с меньшим коэффициентом, остаток заменяем параметром и подставляем обратно.

Например: Решим в целых числах $2x + 3y = 5$: 
1. Выразим $2x = 5 - 3y$.
2. Заметим, что левая часть ($2x$) чётная, значит, правая ($5 - 3y$) тоже должна быть чётной. Это значит, что $3y$ обязано быть нечётным, то есть $y$ — нечётное число ($y = 2k + 1$).
3. Подставляем обратно: $2x = 5 - 3(2k + 1) = 5 - 6k - 3 = 2 - 6k \implies x = 1 - 3k$.
4. Ответ: $x = 1 - 3k, \, y = 2k + 1$ ($k \in \mathbb{Z}$).

## Нелинейные диофантовы уравнения: Метод разложения на множители (Группировка)

Суть: Если в уравнении перемножаются переменные (например, $xy$), стандартный перебор не работает. Трюк заключается в том, чтобы искусственно сгруппировать слагаемые так, чтобы левая часть разложилась на скобки, а в правой осталось число. Затем перебираются все делители этого числа.

Например: Решите в целых числах $xy - 2x - 3y = 1$.
1. группируем первые два слагаемого, вынесем $x$: $x(y - 2) - 3y = 1$.
2. Чтобы получить такую же скобку $(y - 2)$, прибавим и вычтем внутри левой части нужное число: допишем $-6$, тогда сгруппируется как $-3(y - 2) - 6$.Распишем аккуратно: $x(y - 2) - 3(y - 2) - 6 = 1 \implies (x - 3)(y - 2) = 7$.
3. Число $7$ простое, оно раскладывается на пары множителей: $(\pm 1, \pm 7)$ или $(\pm 7, \pm 1)$. Приравниваем каждую скобку к делителям числа $7$ и находим $x, y$.

## Разбор базовых идей

Всегда выбирайте ту переменную, у которой коэффициент по модулю меньше. Если у вас уравнение $7x + 19y = 100$, выражайте $x$: $7x = 100 - 19y$. При делении $19$ на $7$ остаток равен $5$, что сужает перебор остатков до минимума, в отличие от выражения $y$.

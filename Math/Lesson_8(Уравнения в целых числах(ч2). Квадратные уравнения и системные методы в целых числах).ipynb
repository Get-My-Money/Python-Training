## Метод полного квадрата и неравенств (Ограничение области)

Суть: Если уравнение содержит суммы квадратов (например, $x^2 + y^2 = x + y + c$), его можно переписать в виде суммы полных квадратов: $(x - a)^2 + (y - b)^2 = R^2$. Это дает четкую геометрическую интерпретацию (точки внутри или на окружности) и сводит перебор к очень ограниченному числу вариантов.

Пример: Решим в целых числах $x^2 + y^2 = 2x + 2y$.
Перенесем всё влево и выделим квадраты: $(x^2 - 2x + 1) + (y^2 - 2y + 1) = 2 \implies (x - 1)^2 + (y - 1)^2 = 2$.
Сумма двух квадратов целых чисел равна $2$ только если оба квадрата равны $1$:$(x - 1)^2 = 1 \implies x \in \{0, 2\}$ и $(y - 1)^2 = 1 \implies y \in \{0, 2\}$.
Получаем $4$ пары: $(0; 0), (0; 2), (2; 0), (2; 2)$.

## Симметрия и «Оценки» в целых и натуральных числах

Суть: Если в уравнении/системе переменные входят симметрично или заданы ограничения (например, $10$ натуральных чисел в сумме дают $148$), то упорядочение переменных ($x_1 \le x_2 \le \dots \le x_n$) и оценка наименьшего/наибольшего элемента позволяют сузить перебор до $1$–$2$ вариантов.

Пример: Допустим, $a, b, c \in \mathbb{N}$ и $abc = a + b + c$.
Предположим $a \le b \le c$. Разделим на $c$: $ab = \frac{a}{c} + \frac{b}{c} + 1 \le 1 + 1 + 1 = 3$.
Следовательно, $ab \le 3$, откуда легко перебираются пары для $(a, b)$.

## Анализ дробно-рациональных и кубических систем

Суть: При наличии степеней $n^3$ или кубов в уравнении (например, $x^6 = y^3 + k$) используется разность кубов или разность квадратов: $(x^2)^3 - y^3 = k \implies (x^2 - y)(x^4 + x^2 y + y^2) = k$.

# Урок 7. Уравнения в целых числах: Чётность, разложение и метод разности квадратов

## Лемма о чётности суммы и разности

Суть: Если $a$ и $b$ — целые числа, то их сумма $(a + b)$ и разность $(a - b)$ всегда имеют одинаковую чётность (либо обе чётные, либо обе нечётные).

Почему это работает: Заметим, что $(a + b) - (a - b) = 2b$ (всегда чётное число). Разность двух чисел чётна тогда и только тогда, когда сами числа имеют одинаковую чётность.

Например: В уравнении $x^2 - y^2 = 12$, расписанном как $(x - y)(x + y) = 12$, обе скобки $(x - y)$ и $(x + y)$ обязаны иметь одинаковую чётность. Поскольку их произведение чётно ($12$), обе скобки должны быть чётными. Это сразу отсекает пары делителей вроде $(1, 12)$ или $(3, 4)$, оставляя только чётные пары $(\pm 2, \pm 6)$.

## Метод разности квадратов

Суть: Квадратные диофантовы уравнения вида $x^2 - y^2 = c$ или $x^2 = y^2 + ay + b$ сводятся к искусственному выделению полных квадратов. Перенедя всё в одну часть, мы получаем разность квадратов $A^2 - B^2 = c$, которая раскладывается в произведение $(A - B)(A + B) = c$.

Пример: Решим $x^2 = y^2 + 2y + 13$.
Выделим полный квадрат в правой части: $x^2 = (y^2 + 2y + 1) + 12 \implies x^2 - (y + 1)^2 = 12$.
Разложим на множители: $(x - y - 1)(x + y + 1) = 12$. Так как суммы и разности скобок дают чётные числа, обе скобки должны быть чётными делителями числа $12$.

## Разбор базовых идей

Когда уравнение распадается на произведение двух множителей, равных числу $C$ (например, $(A)(B) = C$), люди часто начинают перебирать все делители числа $C$. Олимпиадный навык — сразу проверить чётность суммы этих скобок. Если сумма выражений внутри скобок выражается через $2x$ (чётное число), то сами скобки обязаны быть одинаковой чётности. Если $C$ — нечётное число, решений вообще нет; если чётное — перебираем только чётные пары делителей.

Уравнения вроде $6m^2 - 2n^2 + mn = 3$ кажутся страшными, но они однородны относительно старших степеней или легко раскладываются, если рассматривать их как квадратные относительно одной переменной (например, $m$). Приравняв к нулю и найдя корни через дискриминант, можно разложить левую часть на линейные множители с целыми коэффициентами.

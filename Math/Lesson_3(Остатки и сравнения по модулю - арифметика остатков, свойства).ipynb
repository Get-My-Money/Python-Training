# Урок 3. Остатки и сравнения по модулю

## Деление с остатком

Определение: Разделить целое число $a$ на целое число $b \neq 0$ с остатком — значит представить его в виде:$$a = b \cdot q + r$$где $q$ — неполное частное, а $r$ — остаток, который обязан удовлетворять условию $0 \le r < \vert{}b\vert{}$.

Пример: $23 = 5 \cdot 4 + 3$ (остаток $3$). Для отрицательных чисел: $-23 = 5 \cdot (-5) + 2$ (остаток $2$). В программировании на Python эта логика встроена в оператор %, где -23 % 5 вернёт именно $2$, а не $-3$.

## Сравнения по модулю (Арифметика остатков)

Определение: Говорят, что числа $a$ и $b$ сравнимы по модулю $m$, если они дают одинаковые остатки при делении на $m$. Обозначается:$$a \equiv b \pmod m$$Равносильное условие: разность $(a - b)$ делится на $m$.

Например: $17 \equiv 5 \pmod{12}$, так как $17 - 5 = 12 \vdots 12$. (Это цикличность обычного циферблата часов).

## Свойства сравнений

С остатками можно работать почти как с обычными уравнениями: их можно складывать, вычитать и умножать.

Если $a \equiv r_1 \pmod m$ и $b \equiv r_2 \pmod m$, то:

1. $a \pm b \equiv r_1 \pm r_2 \pmod m$

2. $a \cdot b \equiv r_1 \cdot r_2 \pmod m$

3. $a^k \equiv r_1^k \pmod m$

Например: Найдем остаток от деления $12 \cdot 15$ на $7$. Так как $12 \equiv 5 \pmod 7$ и $15 \equiv 1 \pmod 7$, то $12 \cdot 15 \equiv 5 \cdot 1 = 5 \pmod 7$.

## Решения задач из практикума

1. Десятичная запись числа $M$ состоит из $75$ цифр: только пятёрок, шестёрок и семёрок. Известно, что количество пятёрок на $19$ больше, чем семёрок. Найдите остаток от деления $M$ на $9$.

2. Может ли число вида $100\dots007$ быть квадратом целого числа?

3. Натуральное число $N$ при делении на $15$ даёт неполное частное $x$ и остаток $y$, а при делении на $11$ — неполное частное $y$ и остаток $x$. Найдите все возможные значения $N$.

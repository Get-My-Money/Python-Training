# Урок 4. Остатки и сравнения: периодичность, степени и остатки

## Периодичность остатков степеней

Суть: Последовательность остатков от деления степеней $a^1, a^2, a^3, \dots, a^n$ на фиксированное число $m$ неизбежно зацикливается (становится периодической). Это происходит потому, что возможных остатков по модулю $m$ конечное число.

Пример: Найдем остатки $3^n$ при делении на $5$:$$3^1 \equiv 3, \quad 3^2 \equiv 4, \quad 3^3 \equiv 2, \quad 3^4 \equiv 1, \quad 3^5 \equiv 3 \pmod 5$$Остатки циклически повторяются с периодом $4$: $(3, 4, 2, 1)$. Чтобы найти остаток $3^{102} \pmod 5$, найдем остаток показателя от деления на период: $102 = 4 \cdot 25 + 2 \implies 3^{102} \equiv 3^2 = 9 \equiv 4 \pmod 5$.

## Использование отрицательных остатков

Суть: При возведении в большую степень удобно заменять основание на ближайшее кратное модулю число, переходя к остатку $-1$ (или с малым модулем). Нам помогает тождество: $(-1)^n = 1$ для чётных $n$ и $(-1)^n = -1$ для нечётных $n$.

Например: Найдем остаток от деления $31^{41}$ на $32$. Так как $31 \equiv -1 \pmod{32}$, получаем:$$31^{41} \equiv (-1)^{41} = -1 \equiv 31 \pmod{32}$$

## Системы остатков с постоянным сдвигом

Суть: Если число $N$ при делении на $m_1, m_2, m_3$ даёт остатки $r_1, r_2, r_3$, причём разности $(m_1 - r_1) = (m_2 - r_2) = (m_3 - r_3) = k$ одинаковы, это означает, что числу $N$ до деления нацело во всех случаях «не хватает» ровно $k$ единиц.

Следовательно, $N + k$ делится нацело на $m_1, m_2$ и $m_3$, а значит, $N + k \vdots \text{НОК}(m_1, m_2, m_3)$.

Например: Число $x$ даёт остаток $4$ при делении на $5$ и остаток $6$ при делении на $7$. До каждого делителя не хватает $1$ ($5-4=1$; $7-6=1$). Значит, $x + 1$ делится на $35$, то есть $x = 35m - 1$ (наименьшее такое натуральное число — $34$).

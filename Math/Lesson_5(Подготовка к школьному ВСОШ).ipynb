## Домашнее задания

1. Коля заметил, что для краткой записи дней недели: пн, вт, ср, чт, пт, сб, вс
используются 8 букв, которые встречаются: «б» – 1 раз, «в» – 2 раза, «н» – 1 раз, «п» – 2 раза,
«р» – 1 раз, «с» – 3 раза, «т» – 3 раза, «ч» – 1 раз. Коля выбрал 32 последовательных дня и
для них сосчитал А – количество букв «т», и Б – количество букв «р», встречавшихся в записи
дней недели в выбранный период. Какое наибольшее значение могла принять разность А - Б?

2. Дан квадратный трёхчлен f(x). Известно, что линейная функция y = f(x + 1) − f(x) обращается в ноль при x = 5. При каком значении аргумента обращается в ноль функция y = f(x + 3) − f(x)?

3. Найдите наименьшее число, начинающееся с цифр 11224 и делящееся на 225.

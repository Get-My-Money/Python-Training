# Урок 9. Задачи с целыми числами: неравенства, оценки и текстовые модели

## Целочисленная делимость через проценты и дроби

Суть: Если в задаче сказано, что некоторая часть людей, предметов или чисел составляет $p\%$ (или дробь $\frac{a}{b}$) от общего количества $N$, это скрытое условие делимости. Алгоритм: Переводим проценты в несократимую дробь $\frac{p}{100} = \frac{a}{b}$. Поскольку количество объектов — строго целое число, общее количество $N$ обязано делиться на знаменатель $b$ ($N = b \cdot k$).

Например: В классе $48\%$ девочек. Переведем в дробь: $\frac{48}{100} = \frac{12}{25}$. Значит, общее число учеников в классе кратно $25$ (например, $25, 50, \dots$). Если учеников меньше $40$, то в классе ровно $25$ человек, из них $12$ девочек.

## Суммы последовательных чисел и «выпавший элемент»

Суть: Сумма $n$ последовательных натуральных чисел, начиная с $a$, вычисляется как арифметическая прогрессия:$$S_n = \frac{2a + n - 1}{2} \cdot n$$Если из этой последовательности стёрли одно число $x$ (где $a \le x \le a + n - 1$), то новая сумма $S'$ зажата в строгих границах:$$S_n - (a + n - 1) \le S' \le S_n - a$$Микропример: Написано $10$ последовательных чисел, одно стёрли, сумма оставшихся $50$. Сумма всех $10$ чисел $S_{10} = 5(2a + 9)$ оканчивается на $0$ или $5$. Значит, $S_{10} \in \{55, 60\}$. Проверив оба значения через неравенства для стёртого числа, находим единственное $a$.

## Системы уравнений и неравенств («Метод зажатия»)

Суть: Когда неизвестных больше, чем уравнений (например, $2$ переменные и $1$ уравнение), недостающие связи заменяют неравенствами из здравого смысла: количество предметов $> 0$, доля меньше $100\%$, наименьшее число не больше среднего и т.д. Так как переменные целые, в узком интервале неравенства помещается всего $1$–$2$ числа.

## Разбор базовых идей

Задача: На доске записано несколько последовательных натуральных чисел. Ровно $52\%$ из них — чётные. Сколько чётных чисел записано на доске?

Задача: В ролевой игре $32$ игрока (одиночный или двойной класс: воин, маг, целитель). Целителей в $2$ раза больше магов и в $k$ раз меньше воинов ($k \in \mathbb{N}, k > 2$). Игроков с двойным классом на $2$ больше, чем целителей. Сколько игроков имеют одиночный класс?

## Домашнее задания

1. Алиса купила открытку дешевле $100$ руб. (целое число). Через месяц открытка подорожала в $1.2$ раза, при этом продавец лишь переставил цифры на ценнике. Сколько стоила открытка изначально?

2. Сумма $10$ различных натуральных чисел равна $148$. Наименьшее из них составляет не менее половины от наибольшего. Найдите эти числа.

3. Джек Воробей разложил $150$ пиастров по $10$ кошелькам. В каждый следующий он клал больше, чем в предыдущий. Количество в первом кошельке не меньше половины от последнего. Сколько пиастров в $6$-м кошельке?

4. В школе лодырей $75\%$ учеников не явились на соревнования. В обоих соревнованиях (списывание и подсказка) участвовало $10\%$ всех явившихся. Участвовавших в подсказке было в $1.5$ раза больше, чем в списывании. Найдите наименьшее число учеников в школе.

5. Пятиногие головастики имеют по $1$ хвосту, а четырёхногие — по $k$ хвостов ($k > 1$). Всего поймали головастиков со $100$ ногами и $64$ хвостами. Сколько хвостов у многохвостого головастика?

6. Для укладки пола в квадратной комнате купили плитки. $15$ плиток разбились. Оставшимися выложили прямоугольную комнату, длина которой на $11$ плиток больше ширины. Сколько плиток купили?

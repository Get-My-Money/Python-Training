# Урок 1. Делимость чисел и десятичная запись

## Делимость и разложение на простые множители

Определение: Запись $a \vdots b$ означает, что $a$ делится на $b$ нацело, то есть $a = b \cdot k$, где $k \in \mathbb{Z}$. 

Пример: $15 \vdots 3$, так как $15 = 3 \cdot 5$.

Основная теорема арифметики (ОТА): Любое натуральное число $n > 1$ раскладывается в произведение простых множителей единственным образом:$$n = p_1^{a_1} \cdot p_2^{a_2} \dots p_k^{a_k}$$Например: $360 = 2^3 \cdot 3^2 \cdot 5^1$.

## Количество всех делителей числа

Формула: Если $n = p_1^{a_1} \cdot p_2^{a_2} \dots$, то общее число его натуральных делителей равно:$$d(n) = (a_1 + 1)(a_2 + 1)\dots$$Нужно для того чтобы быстро узнать количество делителей, не выписывая их вручную.

Пример: Посчитаем количество делителей числа $24 = 2^3 \cdot 3^1$.

Берем показатели степеней ($3$ и $1$), прибавляем к каждому по единице и перемножаем:$$d(24) = (3 + 1) \cdot (1 + 1) = 4 \cdot 2 = 8$$

## Степень вхождения простого числа в факториал(Формула Лежандра)

Формула: Показатель степени, с которым простое число $p$ входит в разложение $n!$, равен:$$v_p(n!) = \left\lfloor \frac{n}{p} \right\rfloor + \left\lfloor \frac{n}{p^2} \right\rfloor + \left\lfloor \frac{n}{p^3} \right\rfloor + \dots$$(где $\lfloor x \rfloor$ — целая часть числа, отбрасывание дроби).

Нужно для того чтобы узнать, на какую максимальную степень $p^k$ делится огромное число $n!$.

Например: На какую максимальную степень тройки делится $20!$?

$\lfloor 20 / 3 \rfloor = 6$ (числа $3, 6, 9, 12, 15, 18$ дают по первой тройке).

$\lfloor 20 / 9 \rfloor = 2$ (числа $9, 18$ дают ещё по одной дополнительной тройке).

$\lfloor 20 / 27 \rfloor = 0$ (степени больше $20$ дают $0$).

Складываем: $v_3(20!) = 6 + 2 = 8$. Значит, $20!$ делится на $3^8$, но не делится на $3^9$.

## Десятичная запись и группировка по блокам

Поразрядное разложение: Запись $\overline{abc}$ с чертой означает число, а не произведение $a \cdot b \cdot c$:$$\overline{abc} = 100a + 10b + c$$Пример: $\overline{735} = 7 \cdot 100 + 3 \cdot 10 + 5$.

Разложение блоками (гранями): Четырёхзначное число можно разбить на две пары цифр:$$\overline{abcd} = 100 \cdot \overline{ab} + \overline{cd} = 99 \cdot \overline{ab} + (\overline{ab} + \overline{cd})$$Например: Запишем $4215$ как $4200 + 15 = 100 \cdot 42 + 15 = 99 \cdot 42 + (42 + 15)$.Так как $99 \cdot 42$ точно делится на $99$, остаток от деления $4215$ на $99$ зависит только от суммы пар $42 + 15 = 57$

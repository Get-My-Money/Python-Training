Lesson 28
Мини-задача: «Управляемый квадратик»

```python
import tkinter as tk
```

```python
width = 500
height = 400
size = 30
step = 10

root = tk.Tk()
root.title("Квадрат")

canvas = tk.Canvas(root, width=width, height=height, bg="white")
canvas.pack()


x = width // 2
y = height // 2


square = canvas.create_rectangle(x, y, x + size, y + size, fill="purple")

def move_left(event):
    global x
    if x - step >= 0:
        x -= step
        canvas.coords(square, x, y, x + size, y + size)
        

def move_right(event):
    global x
    if x + step + size <= width:
        x += step
        canvas.coords(square, x, y, x + size, y + size)
        

def move_up(event):
    global y
    if y - step >= 0:
        y -= step
        canvas.coords(square, x, y, x + size, y + size)
        

def move_down(event):
    global y
    if y + step + size <= height:
        y += step
        canvas.coords(square, x, y, x + size, y + size)

        
root.bind("<Left>", move_left)
root.bind("<Right>", move_right)
root.bind("<Up>", move_up)
root.bind("<Down>", move_down)

root.mainloop()
```

Lesson 29
Задача С ― «Автодвижение с ускорением»

```python
width = 500
height = 400
radius = 20
limit = 15

root = tk.Tk()
root.title("Шарик")

canvas = tk.Canvas(root, width=width, height=height, bg="white")
canvas.pack()

x = radius + 50
y = height // 2
speed = 1
direction = 1

ball = canvas.create_oval(x - radius, y - radius, x + radius, y + radius, fill="red")

def update():
    global x, speed, direction
    
    if speed < limit:
        speed += 0.2
    
    x = x + speed * direction
    
    if x + radius >= width:
        x = width - radius
        direction = -1
    
    if x - radius <= 0:
        x = radius
        direction = 1
    
    canvas.coords(ball, x - radius, y - radius, x + radius, y + radius)
    
    root.after(20, update)

update()
root.mainloop()
```

## Задача A ― «Пропущенное число»

Условие: 

    Дан массив чисел от 1 до 100. Но одно число случайно удалили. Как найти его, если массив не отсортирован?

```python
import random

s = [i+1 for i in range(100)]
a = random.randint(1,100)
del s[a-1]

new_sum = sum(s)
old_sum = (1+100)/2*100

print(a, old_sum-new_sum)
```

## Задача B ― «Минимальный обмен»

Условие:

    1. Есть строка состоящая из нулей и единиц типа: "1010011110000000000001"
    
    2. Нужно найти минимальное число замен(свапов), чтобы все единицы стояли подряд.

```python
sts = "00011100111110000000001111111111000111000000"
counter = 0
max_counter = 0
sub_sts = 0

for i in sts:
    if i == "1":
        counter+=1
        
for i in range(0,counter):
    if sts[i] == "1":
        max_counter +=1
sub_sts = max_counter

for i in range(counter, len(sts)):
    sub_sts += int(sts[i]) - int(sts[i-counter])
    if sub_sts > max_counter:
        max_counter = sub_sts
        
print(counter - max_counter)
```

## Задача D ― «Сжатие данных»

Условие: 

    Дана строка типа AAAABBBCCDAA. Нужно превратить её в A4B3C2D1A2.

```python
sts = "AAAABBBCCDAA"
sp = []
counter = 1 
for i in range(len(sts)-1):
    if sts[i] == sts[i+1]:
        counter += 1
    else:
        sp.append(sts[i])
        sp.append(str(counter))
        counter = 1
sp.append(sts[-1])
sp.append(str(counter))
print("".join(sp))
```

## Задача E ― «Анаграммы»

Условие: 

    Даны две строки. Нужно проверить, можно ли из букв первой строки собрать вторую (например: "сон" и "нос" — да, "кабан" и "банка" — да).

```python
first = "кабан"
second = "банка"

first_slovarb = {}
second_slovarb = {}

if len(first) == len(second):

    for i in range(len(first)):
        if first[i] not in first_slovarb:
            first_slovarb[first[i]] = 1
        else:
            first_slovarb[first[i]] += 1
            
    for i in range(len(second)):
        
        if second[i] not in second_slovarb:
            second_slovarb[second[i]] = 1
        else:
            second_slovarb[second[i]] += 1
    print(first_slovarb, second_slovarb)
    if second_slovarb == first_slovarb:
        print("Можно")
    else:
        print("Нельзя")
else:
    print("Нельзя")
```

## Простое отслеживание клавиш

```python
def press_a():
    print("Нажата клавиша a")

keyboard.on_press_key("a", lambda e: press_a())

print("Нажми ESC для выхода")
keyboard.wait("esc")

keyboard.unhook_all()
print("Программа завершена")
```

## keyboard.write() — имитация печати

```python
def bind_Hello_on_b():
    #keyboard.write("Hello")
    keyboard.write("Hello slowly", delay=0.2)

keyboard.on_press_key("b", lambda e: bind_Hello_on_b())

keyboard.wait("esc")
keyboard.unhook_all()
```

## keyboard.press() и keyboard.release()

```python
# press() — нажимает и держит, release() — отпускает
def bind_aaaaa_on_a():
    keyboard.press("a")
    time.sleep(2)
    keyboard.release("a")
    

keyboard.on_press_key("g", lambda e: bind_aaaaa_on_a())
keyboard.wait("esc")
```

```python
time.sleep(2)
keyboard.press("shift")
keyboard.write("hello")
keyboard.release("shift")
```

```python
time.sleep(2)
for char in "hello":
    keyboard.press("shift")
    keyboard.press(char)
    keyboard.release(char)
    keyboard.release("shift")
    time.sleep(0.1)
```

## keyboard.add_hotkey()

```python
def hello():
    print("Привет!")

keyboard.add_hotkey("ctrl+g", hello)

keyboard.wait("esc")
keyboard.unhook_all()
```

## keyboard.record() и keyboard.play()

```python
def record_actions():
    print("Запись... нажми L для остановки")
    events = keyboard.record("l")         
    print("Воспроизведение...")
    keyboard.play(events)
    # keyboard.play(events, speed_factor=2) - ускорение

def start_recording():
    t = threading.Thread(target=record_actions)
    t.start()

keyboard.add_hotkey("ctrl+g", start_recording)

print("Нажми ESC для выхода")
keyboard.wait("esc")
keyboard.unhook_all()
print("Программа завершена")
```

## Комбинированный пример макроса

```python
def macro():
    keyboard.write("Hello")
    time.sleep(2)
    keyboard.press_and_release("enter")

keyboard.add_hotkey("ctrl+b", macro)

keyboard.wait("esc")
keyboard.unhook_all()
```

## «Мини-макрос система»

Реализовать:

    1. Макросы:
    
        Ctrl + 1 → пишет "Привет!"
        
        Ctrl + 2 → пишет "Как дела?"
        
        Ctrl + 3 → пишет текущее время (можно через datetime)
        
        Ctrl + 4 → делает “Enter + текст”

    2. Анти-спам: нельзя нажимать макрос чаще чем раз в 1 секунду

    3. Свой мини-лог: [12:30:15] Использован макрос: hello

    4. Параметры выхода: esc должен остановить программу и удалить все бинды

```python
info_log = {}

def hello():
    a = str(dt.datetime.now()).split()[1]
    print("Привет!")
    info_log[a] = "Hello" 
    time.sleep(1)
keyboard.add_hotkey("ctrl+1", hello)

def hello2():
    a = str(dt.datetime.now()).split()[1]
    print("Как дела?")
    info_log[a] = "How are you?" 
    time.sleep(1)
keyboard.add_hotkey("ctrl+2", hello2)


def hello3():
    a = str(dt.datetime.now()).split()[1]
    print(f"Текущее время: {a}")
    info_log[a] = "What's time now" 
    time.sleep(1)
keyboard.add_hotkey("ctrl+3", hello3)

def hello4():
    a = str(dt.datetime.now()).split()[1]
    keyboard.write("Hello")
    keyboard.press_and_release("enter")
    info_log[a] = "Enter+Text" 
    time.sleep(1)
    
keyboard.add_hotkey("ctrl+4", hello4)


print("Нажми ESC для выхода")
keyboard.wait("esc")
keyboard.unhook_all()
print("Программа завершена")
```

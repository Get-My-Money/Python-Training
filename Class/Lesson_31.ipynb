## Задача A ― «Злой квадрат»

Условие:

    1. Создать игрока (квадрат), врага (квадрат другого цвета)
    2. Игрок управляется стрелками
    3. Враг в game_loop двигается к игроку
    4. Столкновение: если враг касается игрока ― GAME OVER

```python
width = 500
height = 400
size = 25
step = 10

root = tk.Tk()
root.title("Game: Злой квадрат")

canvas = tk.Canvas(root, width=width, height=height, bg="white")
canvas.pack()

x = width // 1.3
y = height // 1.3

x2 = width // 6
y2 = height // 6
square_me = canvas.create_rectangle(x, y, x + size, y + size, fill="green")
square_enemy = canvas.create_rectangle(x2, y2, x2 + size, y2 + size, fill="red")

def enemy_moving():
    global x2, y2
    # Получаем координаты игрока
    player_coords = canvas.coords(square_me)
    player_x = player_coords[0]
    player_y = player_coords[1]
    
    enemy_coords = canvas.coords(square_enemy)
    enemy_x = enemy_coords[0]
    enemy_y = enemy_coords[1]
    if player_x == enemy_x or player_y == enemy_y:
        print("Game over")
        canvas.delete(square_enemy)
        canvas.delete(square_me)
        canvas.create_text(250, 150, text="Game over", fill = "red")
        return None
    
    
    # Вычисляем вектор от врага к игроку
    dx = player_x - x2
    dy = player_y - y2
    distance = (dx**2 + dy**2)**0.5
    
    speed = 5
    if distance > speed:
        step_x = dx / distance * speed
        step_y = dy / distance * speed
    else:
        step_x = dx
        step_y = dy   
    x2 += step_x
    y2 += step_y
        
    # Обновляем положение врага на холсте
    canvas.coords(square_enemy, x2, y2, x2 + size, y2 + size)
    
    # Планируем следующий шаг через 50 мс (20 раз в секунду)
    root.after(50, enemy_moving)

def move_left(event):
    global x
    if x - step >= 0:
        x -= step
        canvas.coords(square_me, x, y, x + size, y + size)
    else:
        x = 3
        canvas.coords(square_me, x, y, x + size, y + size)

def move_right(event):
    global x
    if x + step + size <= width:
        x += step
        canvas.coords(square_me, x, y, x + size, y + size)
    else:
        x = width - size - 3
        canvas.coords(square_me, x, y, x + size, y + size)
        


def move_up(event):
    global y
    if y - step >= 0:
        y -= step
        canvas.coords(square_me, x, y, x + size, y + size)
    else:
        y = 3
        canvas.coords(square_me, x, y, x + size, y + size)



def move_down(event):
    global y
    if y + step + size <= height:
        y += step
        canvas.coords(square_me, x, y, x + size, y + size)
    else:
        y = height - size - 3
        canvas.coords(square_me, x, y, x + size, y + size)

enemy_moving()
        
root.bind("<Left>", move_left)
root.bind("<Right>", move_right)
root.bind("<Up>", move_up)
root.bind("<Down>", move_down)
root.mainloop()
```

# 1. Введение

## Цель проекта

Разработать десктопное приложение для хранения и систематизации олимпиадных задач по программированию.

Программа должна позволять:

- хранить условия задач;
- сохранять решения;
- распределять задачи по темам;
- отслеживать прогресс решения задач;
- анализировать статистику подготовки.

---

# 2. Архитектура и стек технологий

Проект разрабатывается на Python с использованием объектно-ориентированного подхода.

## Используемые технологии

| Компонент | Технология |
|---|---|
| Язык программирования | Python 3.x |
| Интерфейс | Tkinter |
| Работа с файлами | os, pathlib |
| Хранение данных | JSON |
| Аналитика | matplotlib, datetime |

---

# Структура проекта

Все данные приложения должны храниться внутри папки:

```text
data/
```

Пример структуры:

```text
data/
└── arrays/
    └── two_sum/
        ├── meta.json
        ├── condition.txt
        └── solution.py
```

---

# Назначение файлов

| Файл | Назначение |
|---|---|
| meta.json | Информация о задаче |
| condition.txt | Текст условия |
| solution.py | Код решения |

---

# 3. Формат хранения данных

Информация о каждой задаче хранится в JSON-файле.

Пример:

```json
{
  "name": "Two Sum",
  "topic": "Arrays",
  "difficulty": easy,
  "status": "Solved",
  "date": "2026-05-15 12:00"
}
```

---

# Описание полей

| Поле | Тип | Описание |
|---|---|---|
| name | string | Название задачи |
| topic | string | Тема задачи |
| difficulty | string | Сложность easy/medium/hard |
| status | string | Статус решения |
| date | string | Дата добавления |

---

# 4. Функциональные требования

# 4.1 Backend

Необходимо реализовать класс:

```python
TaskManager
```

Класс должен отвечать за:

- создание папки `data`, если она отсутствует;
- загрузку списка задач при запуске программы;
- сохранение новых задач;
- изменение статуса задачи;
- удаление задач.

---

# Основные методы

## save_task()

Метод должен:

- создавать папку задачи;
- сохранять:
  - meta.json
  - condition.txt
  - solution.py

---

## update_status()

Метод обновляет статус задачи внутри файла:

```text
meta.json
```

---

## load_tasks()

Метод должен:

- обходить директорию `data`;
- загружать список всех задач;
- формировать структуру данных для интерфейса.

---

# 4.2 Графический интерфейс

Интерфейс должен быть разделен на несколько блоков.

Для построения интерфейса использовать:

```python
tkinter.Frame
```

---

# Основные элементы интерфейса

| Элемент | Виджет | Назначение |

---

# Основные кнопки

Интерфейс должен содержать кнопки:

- Добавить задачу
- Показать решение
- Удалить задачу
- Изменить статус

---

# 5. Аналитика и статистика

Программа должна вычислять процент решенных задач для каждой темы.

Формула:

```text
P(T) = (Solved / Total) * 100%
```

где:

- Solved — количество решенных задач;
- Total — общее количество задач.
